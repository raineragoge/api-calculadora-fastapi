```python

import requests

# URL pública de nuestra API
url = "https://gown-proximity-aggregate.ngrok-free.dev/doble/12"

# Realizar petición GET
respuesta = requests.get(
    url,
    headers={"ngrok-skip-browser-warning": "true"},
    timeout=15
)

# Mostrar respuesta
print("Código HTTP:", respuesta.status_code)
print("Resultado:", respuesta.json())
```

```python

import requests

url = "https://gown-proximity-aggregate.ngrok-free.dev/sumar"

# Datos que enviaremos a la API
datos = {
    "a": 35,
    "b": 65
}

# Petición POST
respuesta = requests.post(
    url,
    json=datos,
    headers={"ngrok-skip-browser-warning": "true"},
    timeout=15
)

print("Código HTTP:", respuesta.status_code)
print("Resultado:", respuesta.json())
```
